## 3. Inspeção das Tabelas Fato e Dimensão Geradas 

```python
from pathlib import Path
from IPython.display import display
import pandas as pd

# Caminho da pasta onde estão os arquivos Parquet
processed_dir = Path("/content/projeto_aneel_equipe14/data/processed")

# Lista todos os arquivos .parquet da pasta
parquet_files = sorted(processed_dir.glob("*.parquet"))

if not parquet_files:
    print(f"Nenhum arquivo Parquet encontrado em: {processed_dir}")

for file_path in parquet_files:
    print("=" * 80)
    print("\n")
    print(f"Tabela: {file_path.name}")

    # Lê apenas o necessário ou o arquivo inteiro
    df = pd.read_parquet(file_path)

    print(f"Dimensões: {df.shape[0]:,} linhas x {df.shape[1]} colunas")
    display(df.head(5))
```
